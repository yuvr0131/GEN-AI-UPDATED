```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize
nltk.download('punkt_tab')
#sample text for tokenization
text="NLTK provides powerful tool for tokenization.It includes word tokenization and sentence tokenization"
```

```python
#word tokenization
words=word_tokenize(text)
print(words)
```

```python
#sentence tokenization
sent=sent_tokenize(text)
print(sent)
```

```python
#stopwords removal
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

```

```python
#sentence
sentence="This is a sample sentence, showing off the words filtration"
```

```python
#tokenize the sentence
nltk.download('punkt')
nltk.download('stopwords')
words=word_tokenize(sentence)

#flitering out stopwords
new_sentence=[word for word in words if word.lower() not in stopwords.words('english')]
```

```python
print(sentence)
print(new_sentence)
```
